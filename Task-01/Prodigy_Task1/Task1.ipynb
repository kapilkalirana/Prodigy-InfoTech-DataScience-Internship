```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
population_df = pd.read_csv(
    r"D:\data files\API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv",
    skiprows=4
)

population_df.head()
```

```python
population_df.info()
```

```python
print(population_df.columns.tolist())
```

```python
population = population_df[['Country Name', '2023']]

population = population.dropna()

population.head()
```

```python
country_metadata = pd.read_csv(
    r"D:\data files\Metadata_Country_API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv"
)

country_metadata.head()
```

```python
country_metadata.info()
```

```python
print(country_metadata.columns.tolist())
```

```python
indicator_metadata = pd.read_csv(
    r"D:\data files\Metadata_Indicator_API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv"
)

indicator_metadata.head()
```

```python
indicator_metadata.info()
```

```python
print(indicator_metadata.columns.tolist())
```

```python
merged_data = pd.merge(
    population,
    country_metadata,
    left_on="Country Name",
    right_on="TableName",
    how="left"
)

merged_data.head()
```

```python
plt.figure(figsize=(10,6))

plt.hist(population['2023'], bins=30, edgecolor='black')

plt.title("Distribution of Country Population")

plt.xlabel("Population")

plt.ylabel("Number of Countries")

plt.grid(True)

plt.show()
```

```python
top10 = population.sort_values(
    by='2023',
    ascending=False
).head(10)

top10
```

```python
plt.figure(figsize=(12,6))

plt.bar(top10['Country Name'], top10['2023'])

plt.title("Top 10 Most Populous Countries")

plt.xlabel("Country")

plt.ylabel("Population")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
region_count = merged_data['Region'].value_counts()

region_count
```

```python
plt.figure(figsize=(10,6))

region_count.plot(kind='bar')

plt.title("Countries by Region")

plt.xlabel("Region")

plt.ylabel("Number of Countries")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```
